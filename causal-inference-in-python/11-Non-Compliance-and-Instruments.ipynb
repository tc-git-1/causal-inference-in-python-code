```python
Z = df["prime_elegible"]
T = df["prime_card"]
n = len(df)

# not the same as iv_regr.resid!
e_iv = df["pv"] - iv_regr.predict(df)
compliance = np.cov(T, Z)[0, 1]/Z.var()

se = np.std(e_iv)/(compliance*np.std(Z)*np.sqrt(n))

print("SE IV:", se)
print("95% CI:", [late - 2*se, late + 2*se])
```

### Matrix Implementation


```python
Z = df[["prime_elegible", "credit_score", "income", "age"]].values
X = df[["prime_card", "income", "age"]].values
Y = df[["pv"]].values

def add_intercept(x):
    return np.concatenate([np.ones((x.shape[0], 1)), x], axis=1)

Z_ = add_intercept(Z)
X_ = add_intercept(X)

# pre-multiplying Z_.dot(...) last is important to avoid
# creating a huge NxN matrix
X_hat = Z_.dot(np.linalg.inv(Z_.T.dot(Z_)).dot(Z_.T).dot(X_))

b_iv = np.linalg.inv(X_hat.T.dot(X_hat)).dot(X_hat.T).dot(Y)
b_iv[1]
```

```python
e_hat_iv = (Y - X_.dot(b_iv))

var = e_hat_iv.var()*np.diag(np.linalg.inv(X_hat.T.dot(X_hat)))

np.sqrt(var[1])
```

```python
t_tilde = smf.ols("prime_card ~ income + age", data=df).fit().resid

e_hat_iv.std()/(t_tilde.std()*np.sqrt(n*first_stage.rsquared))
```

### The IV Estimate

Errata: The book does not center the discontinuity at zero. Therefore, the intercept for the ITTE regression cannot be interpreted as the ITTE. Here, I'm centering the discontinuity at zero.

```python
def rdd_iv(data, y, t, r, cutoff):
    
    centered_df = data.assign(**{r: data[r]-cutoff})
    
    compliance = smf.ols(f"{t}~{r}*I({r}>0)", centered_df).fit()
    itte = smf.ols(f"{y}~{r}*I({r}>0)", centered_df).fit()
    
    param = f"I({r} > 0)[T.True]"
    return itte.params[param]/compliance.params[param]


rdd_iv(df_dd, y="pv", t="prime_card", r="balance", cutoff=5000)
```

```python
from joblib import Parallel, delayed
from toolz import partial

def bootstrap(data, est_fn, rounds=200, seed=123, pcts=[2.5, 97.5]):
    np.random.seed(seed)
    
    stats = Parallel(n_jobs=4)(
        delayed(est_fn)(data.sample(frac=1, replace=True))
        for _ in range(rounds)
    )
    
    return np.percentile(stats, pcts)


bootstrap(df_dd,
          partial(rdd_iv, y="pv", t="prime_card", r="balance", cutoff=5000))
```

# Tom's exmaple  
Now the diagram looks like this:  
Birth Quarter → Years of Schooling → Income  
Birth Quarter → Relative Age → Income  


```python
# Tom's example
import numpy as np
import pandas as pd
import statsmodels.api as sm
from linearmodels.iv import IV2SLS  # Install with: pip install linearmodels

# -----------------------------
# Step 1: Simulate Data
# -----------------------------
np.random.seed(42)
n = 2000

# Instrument: Birth Quarter (1 to 4)
birth_quarter = np.random.randint(1, 5, size=n)

# Relative Age: older if born earlier
relative_age = (5 - birth_quarter) + np.random.normal(0, 0.5, size=n)

# Introduce negative correlation between schooling and relative age
schooling = 12 + 0.3 * birth_quarter - 0.8 * relative_age + np.random.normal(0, 1, size=n)

# True model for log income
beta_schooling = 0.12  # 12% per year of schooling
beta_age = 0.02        # 2% per unit of relative age
base = 10.0

log_income = base + beta_schooling * schooling + beta_age * relative_age + np.random.normal(0, 0.05, size=n)
income = np.exp(log_income)

# Create DataFrame
df = pd.DataFrame({
    'income': income,
    'log_income': log_income,
    'schooling': schooling,
    'birth_quarter': birth_quarter,
    'relative_age': relative_age
})

# Display first few rows
print(df.head())
```

```python
# -----------------------------
# Step 2: OLS Models
# -----------------------------

# Model 1: log(income) ~ schooling + birth_quarter
X1 = sm.add_constant(df[['schooling', 'birth_quarter']])
ols1 = sm.OLS(df['log_income'], X1).fit()

print("OLS Model 1 (No Relative Age):")
print(ols1.summary())
print("")

print("INTERPRETATION:\nIn Model 1, birth_quarter (γ) will likely be significant because it affects income via relative age.")
```

```python
# Model 2: log(income) ~ schooling + birth_quarter + relative_age
X2 = sm.add_constant(df[['schooling', 'birth_quarter', 'relative_age']])
ols2 = sm.OLS(df['log_income'], X2).fit()
print("OLS Model 2 (With Relative Age):")
print(ols2.summary())
print("")
print("INTERPRETATION:\nIn Model 2, after adding relative_age, γ should become insignificant, showing that controlling for relative age blocks the extra path.")
```

```python
# -----------------------------
# Step 3: IV Models (2SLS)
# -----------------------------

# IV Model 1: log(income) ~ [schooling ~ birth_quarter]
iv1 = IV2SLS.from_formula('log_income ~ 1 + [schooling ~ birth_quarter]', data=df).fit()
print("\nIV Model 1 (No Relative Age):")
print(iv1.summary)
print("")

print("INTERPRETATION:\nIV Model 1 may be biased because birth quarter affects income through relative age.")
```

```python
# IV Model 2: log(income) ~ relative_age + [schooling ~ birth_quarter]
iv2 = IV2SLS.from_formula('log_income ~ 1 + relative_age + [schooling ~ birth_quarter]', data=df).fit()
print("\nIV Model 2 (With Relative Age):")
print(iv2.summary)
print("")

print("IV Model 2 includes relative age as a control, making the exclusion restriction more plausible.")
```
